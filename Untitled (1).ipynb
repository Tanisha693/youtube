```python
import re
import nltk
from nltk.corpus import stopwords

# Download stopwords (only once)
nltk.download("stopwords")
STOPWORDS = set(stopwords.words("english"))

# Function to clean text
def clean_text(text):
    if not isinstance(text, str):  # if not string, return empty string
        return ""
    text = text.lower()  # convert to lowercase
    text = re.sub(r"http\S+", "", text)  # remove URLs
    text = re.sub(r"[^a-z\s]", "", text)  # keep only letters and spaces
    words = text.split()
    words = [w for w in words if w not in STOPWORDS]  # remove stopwords
    return " ".join(words)

# Make a copy of DataFrame
df = df.copy()

# Apply cleaning to the 'Comment' column
df["clean_comment"] = df["Comment"].apply(clean_text)

# Preview the cleaned text
print(df[["Comment", "clean_comment"]].head(10))
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# Initialize analyzer
analyzer = SentimentIntensityAnalyzer()

# Function to classify sentiment
def sentiment_label(text):
    score = analyzer.polarity_scores(text)["compound"]
    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    else:
        return "Neutral"

# Copy of a df
df = df.copy()

# Sentiment analysis
df.loc[:, "predicted_sentiment"] = df["clean_comment"].apply(sentiment_label)

# Results
print(df[["clean_comment", "predicted_sentiment"]].head(10))
```

```python
# Import Required Libraries
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.svm import LinearSVC
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
from imblearn.over_sampling import SMOTE
import joblib
```

```python
# Load Dataset
df = pd.read_csv("Cleaned_YouTube_Comments.csv")

# Keep only the required columns
df = df[['clean_comment', 'Sentiment']].dropna()

# Features and Labels
X = df['clean_comment'].astype(str)
y = df['Sentiment']
print("COMPLETED SUCCESSFULLY")
```

```python
# Convert Text to Numerical Values
vectorizer = TfidfVectorizer(max_features=8000, ngram_range=(1, 2))
X_vectorized = vectorizer.fit_transform(X)
print("Vectorization Completed")
```

```python
# Balance Data (SMOTE)
smote = SMOTE(random_state=42)
X_balanced, y_balanced = smote.fit_resample(X_vectorized, y)
```

```python
# Split Data into Train & Test Sets
X_train, X_test, y_train, y_test = train_test_split(
    X_balanced, y_balanced, test_size=0.2, random_state=42
)
print("Split Completion")
```

```python
# Train SVM Model
model = LinearSVC()
model.fit(X_train, y_train)
print("Model Trained Succesfully")
```

```python
# Evaluate Model
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print(f"\n Model Accuracy: {accuracy * 100:.2f}%")
print("\n Classification Report:\n", classification_report(y_test, y_pred))
```

```python
logreg_model = LogisticRegression(max_iter=1000, random_state=42)
logreg_model.fit(X_train, y_train)
print("Logistic Regression Model Trained Successfully")
```

```python
# Evaluate Logistic Regression Model
y_pred_logreg = logreg_model.predict(X_test)
accuracy_logreg = accuracy_score(y_test, y_pred_logreg)
print(f"\n Logistic Regression Model Accuracy: {accuracy_logreg * 100:.2f}%")
print("\nLogistic Regression Classification Report:\n", classification_report(y_test, y_pred_logreg))
```

```python
#Train Decision Tree Model
from sklearn.tree import DecisionTreeClassifier
dt_model = DecisionTreeClassifier(random_state=42)
dt_model.fit(X_train, y_train)
y_pred_dt = dt_model.predict(X_test)
accuracy_dt = accuracy_score(y_test, y_pred_dt)
print(f"\n Decision Tree Accuracy: {accuracy_dt * 100:.2f}%")
print(classification_report(y_test, y_pred_dt))
```

```python
import matplotlib.pyplot as plt

# Model accuracies
accuracy_scores = {
    "Logistic Regression": 0.80,
    "Decision Tree": 0.79,
    "SVM": 0.83
}

models = list(accuracy_scores.keys())
scores = list(accuracy_scores.values())

plt.figure(figsize=(8, 5))

# Bar plot
plt.bar(models, scores)

plt.title("Model Accuracy Comparison", fontsize=14, weight='bold')
plt.ylabel("Accuracy Score", fontsize=12)
plt.xlabel("Model", fontsize=12)
plt.ylim(0, 1)

# Add value labels on bars
for i, v in enumerate(scores):
    plt.text(i, v + 0.01, f"{v:.2f}", ha='center', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()
```
